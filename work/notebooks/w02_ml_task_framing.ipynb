## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*



The target is the click-through rate (CTR) or whether a user clicked on a piece of content.

If click labels are available, the model predicts:
- 1 = Clicked
- 0 = Not Clicked

If labels are unavailable, CTR can be used as a proxy target because it measures user engagement.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("./content_refresh_anonymized.csv")

# Preview the target-related columns
df[["ctr", "clicks_90d", "impressions_90d"]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*



The primary evaluation metric is NDCG (Normalized Discounted Cumulative Gain), which measures how well the ranking matches user preferences.

A higher NDCG score indicates better ranking performance.

From a business perspective, success also means improving click-through rate (CTR).

A fixed rule cannot capture these complex relationships.
