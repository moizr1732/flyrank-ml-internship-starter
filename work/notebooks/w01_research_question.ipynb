## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Question: Given a client's existing content inventory, which pages should a content reviewer look at first this week?

Unit of analysis: one content item (content_id) belonging to one client (client_id) - a single published page.

Output: a ranked list of content items, each with a score and a reason code (e.g. declining_with_demand, low_ctr_visible_page) explaining why it surfaced.

Decision it improves: which handful of pages (out of thousands per client) a content team spends its limited weekly review time on, instead of guessing or working alphabetically/by publish date.

Action someone takes: a content reviewer opens the top-ranked pages and either rewrites/expands them, fixes titles or meta descriptions, or confirms the page is fine and moves on.

Cost of a wrong recommendation: two-sided. If the model flags a healthy page as high-priority, the team wastes review hours on something that didn't need it (an opportunity cost - some other page that did need attention doesn't get looked at). If the model misses a genuinely declining, high-impressions page, the client keeps losing visibility/traffic on something that could have been fixed. Because review capacity is small relative to inventory size, false positives near the top of the list are the most expensive mistake - they directly displace a true positive from getting reviewed.

Why data/ML can help at all: with client inventories in the hundreds-to-thousands of pages, a human can't manually re-check every page's trend, position, and CTR every week. A model that reads observable signals and ranks pages by evidence-backed opportunity turns an unmanageable inventory into a short, defensible weekly list.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/moizr1732/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv")

print("Total rows in starter dataset:", len(df))
print("Unique clients:", df['client_id'].nunique())
print()

declining_pct = (df['trend_direction'] == 'down').mean() * 100
print(f"Share of pages trending down (trend_direction == 'down'): {declining_pct:.1f}%")

declining_with_demand = df[(df['trend_direction'] == 'down') & (df['impressions_90d'] >= 100)]
print(f"'declining_with_demand' candidates: {len(declining_with_demand)}")

low_ctr = df[(df['impressions_90d'] >= 500) & (df['avg_position'] > 0) & (df['avg_position'] <= 20) & (df['ctr'] < 0.5)]
print(f"'low_ctr_visible_page' candidates: {len(low_ctr)}")

print()
print("Median impressions_90d:", df['impressions_90d'].median())
print("Median content_age_days:", df['content_age_days'].median())
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) - and what it never will (causal proof, 'predicting Google').*

What I can say: this project observes real, historical search and engagement signals (impressions, clicks, CTR, position, trend, freshness) and ranks content items by how much evidence points to an opportunity or risk. Results are observational and decision-support - they tell a reviewer where to look first, based on patterns in the data, not a guarantee of what will happen if they act.

What I cannot say: I cannot claim this predicts or explains Google's ranking algorithm. I cannot claim that refreshing a flagged page causes a recovery - proving causation would require an actual experiment, which this dataset does not give me. I cannot claim my model captures every real-world reason a page under- or over-performs (seasonality, consolidation, genuine demand drop) unless I explicitly check for those look-alikes. Any current-window label like trend_direction is a proxy, not a future outcome.
