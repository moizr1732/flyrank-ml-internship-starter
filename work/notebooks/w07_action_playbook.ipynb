## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

## Ranked Content Action Playbook

The ranked queue helps prioritize which content should be reviewed first. The recommendations are intended to support editorial decisions rather than automatically change website content.

| Priority | Action | Reason Code | Explanation |
|----------|--------|-------------|-------------|
| 1 | Refresh older high-potential content | RC01 | The model suggests this content may benefit from an update. |
| 2 | Update outdated statistics and references | RC02 | Information may no longer be current. |
| 3 | Improve headings and metadata | RC03 | Small improvements may increase discoverability. |
| 4 | Expand pages with limited information | RC04 | Additional useful content could improve quality. |
| 5 | Monitor only | RC05 | No strong evidence that immediate action is needed. |

### Archetype Mapping

- High score + old content → Refresh first
- High score + recent content → Minor optimization
- Medium score → Human review
- Low score → Monitor
- Very low score → Leave unchanged

The reason codes explain why an item appears in the queue, making the recommendations easier for reviewers to understand.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human Review

Before applying any recommendation, a reviewer should confirm:

- The recommendation matches business goals.
- Facts and references are still accurate.
- Content quality actually needs improvement.
- SEO changes are reasonable.
- The page is still relevant.

## No-Go List

The following actions should never be fully automated:

- Publishing articles
- Deleting pages
- Rewriting content without review
- Medical, legal, or financial advice
- Final editorial approval

A human should always make the final decision before changes are published.

*What would tell you the recommendations went stale?*
## Monitoring

The recommendations should be monitored over time to ensure they remain useful.

Possible indicators include:

- Prediction quality decreases.
- Editorial feedback becomes consistently negative.
- New content types appear.
- Feature distributions change.

## Retraining Triggers

Retraining should be considered when:

- Model performance drops compared with the previous validated version.
- A significant amount of new training data becomes available.
- The website structure changes.

## 5. Exports for the paper

*Write the queue (and any figures you want to reuse) to work/outputs/ — your paper builds on these files.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)

queue = pd.DataFrame({
    "Priority":[1,2,3,4,5],
    "Action":[
        "Refresh content",
        "Update references",
        "Improve metadata",
        "Expand content",
        "Monitor"
    ],
    "Reason_Code":[
        "RC01",
        "RC02",
        "RC03",
        "RC04",
        "RC05"
    ]
})

queue.to_csv(output_dir/"ranked_action_queue.csv", index=False)

metrics = {
    "playbook_version":"1.0",
    "num_actions":len(queue),
    "review_required":True,
    "automation":"Decision Support Only"
}

pd.Series(metrics).to_json(output_dir/"action_summary.json")

plt.figure(figsize=(6,3))
plt.bar(queue["Reason_Code"], queue["Priority"])
plt.title("Content Action Priorities")
plt.xlabel("Reason Code")
plt.ylabel("Priority")
plt.tight_layout()
plt.savefig(output_dir/"action_priority.png")
plt.close()

print("Exports completed successfully.")
```
